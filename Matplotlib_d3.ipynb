```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
Sales_df = pd.read_excel('Sales_2025_20_records.xlsx')
Sales_df
```

```python
month = Sales_df['MonthName']
sales_amount = Sales_df['Sales_Amount']
course = Sales_df['Course_Name']
```

```python
plt.subplot(2,1,1)
plt.plot(month,sales_amount)
plt.subplot(2,1,2)
plt.bar(month,sales_amount)
plt.show()
```

```python
plt.subplot(1,2,1)
plt.plot(month,sales_amount)
plt.subplot(1,2,2)
plt.bar(month,sales_amount)
plt.show()
```

```python
plt.subplot(2,2,1)
plt.barh(course,sales_amount)
plt.subplot(2,2,2)
plt.scatter(month,sales_amount)
plt.subplot(2,2,3)
plt.plot(month,sales_amount)
plt.subplot(2,2,4)
plt.bar(month,sales_amount)
plt.show()
```
