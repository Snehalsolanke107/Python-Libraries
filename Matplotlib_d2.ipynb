```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
sales_df = pd.read_excel('Sales_2025_20_records.xlsx')
sales_df
```

```python
month = sales_df['MonthName']
sales_amount = sales_df['Sales_Amount']
plt.plot(month,sales_amount,marker='o')
```

```python
month_wise_sales = sales_df.groupby('MonthName')['Sales_Amount'].sum()
month_wise_sales
```

```python
month = month_wise_sales.index
Total_sales = month_wise_sales.values
plt.plot(month,Total_sales,color='red',linewidth=2,linestyle='--',marker = 'o',ms=5,mfc='green',mec='black')
plt.title('Sales Record')
plt.xlabel('Months wise sales')
plt.ylabel('Total sales count')
plt.ticklabel_format(axis='y',style='plain')
plt.plot(month,Total_sales)
plt.grid()
plt.show()
```

```python
admission_df = pd.read_excel('students_addmission_records.xlsx')
admission_df
```

```python
course_wise_revenue = admission_df.groupby('Course')['Total Fees'].sum()
course_wise_revenue
```

```python
x = course_wise_revenue.index
y = course_wise_revenue.values
plt.bar(x,y)
```

```python
x = course_wise_revenue.index
y = course_wise_revenue.values
plt.bar(x,y,width=0.5,color='green')
plt.xlabel("Course")
plt.ylabel("Revenue")
plt.title("Course Wise Revenue")
```

```python
Branch_wise_fees = admission_df.groupby('Branch')['Remaining Fees'].sum()
Branch_wise_fees
```

```python
branch = Branch_wise_fees.index
remaining_fees = Branch_wise_fees.values
plt.barh(branch,remaining_fees,color='purple')
plt.xlabel("Remaining fees")
plt.ylabel("Branch")
plt.title("Branch Wise remaining fees")
```

```python
data = admission_df.groupby('Course')['Paid Fees'].sum()
course = data.index 
total_paid = data.values
```

```python
plt.pie(x = total_paid,labels=course,autopct = '%.2f%%')
```
