```python
admission_data = {
    'name': ['vijay', 'ajay', 'sujay', 'jay', 'arun'],
    'course_name': ['Java', 'Python', 'Java', 'Java', 'Python'],
    'fees': [10000, 20000, 10000, 10000, 20000]
    }

ad_df = pd.DataFrame(admission_data)
ad_df
```

```python
def apply_dis(row):
    if row['course_name'] == 'java':
        fees = row['fees'] - row['fees']*10/100
        return fees
    elif row['course_name'] == 'python':
        fees = row['fees'] - row['fees']*20/100
        return fees
    else:
        return row['fees']
```

```python
ad_df['fees_af_dis'] = ad_df.apply(apply_dis,axis = 1)
```
