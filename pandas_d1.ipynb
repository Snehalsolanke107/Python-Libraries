```python
# list compression
employee = {
    'emp_name' : ['jay','bharat','suraj','vijay','Anjali','Rani'],
    'emp_salary' : [40000,45000,50000,30000,35000,60000],
    'emp_dept' : ['IT','HR','Manager','IT','Engineer','HR']
}
emp_df = pd.DataFrame(employee,index=['e' + str(num) for num in range(101,len(employee['emp_name'])+101)])
emp_df
```

```python
# how to access row
#iloc,loc
print(emp_df.iloc[1])
print()
print(emp_df.loc['e102'])
```

```python
emp_df.iloc[1:3]
```

```python
emp_df.loc[['e101','e103']]
```

```python
print(emp_df['emp_dept'])
print()
print(emp_df[['emp_name','emp_salary']])
```

```python
# how to add new column
# df['column name'] = [data]
emp_df['city'] = ['pune','mumbai','nashik','chennai','pune','nagpur']
emp_df
```

```python
emp_df['new_sal'] = emp_df['emp_salary'] + emp_df['emp_increment']
emp_df
```
