```python
import numpy as np
```

```python
# Create the following arrays:
students = np.array(["Amit","Rahul","Priya","Sneha","Vikas","Neha","Rohit","Pooja","Akash","Kiran"])
marks = np.array([78,45,92,67,35,88,56,74,95,62])
print(students)
print(marks)
```

```python
# 1. Number of students 
print(len(students))
# 2. Number of dimensions  
print(marks.ndim)
# 3. Shape of the marks array 
print(marks.shape)
# 4. Size of the marks array  
print(marks.size)
# 5. Data type of the marks array  
print(marks.dtype)
```

```python
# 1. Display marks in ascending order using np.sort(). 
print(np.sort(marks))
print()

# 2. Display marks in descending order. 
print(np.sort(marks)[::-1])
print()

# 3. Sort students according to their marks using np.argsort(). 
ar = np.argsort(marks)[::-1]
for index in ar:
    print(students[index])
print()

# 4. Display the Top 3 students based on marks. 
t3 = ar[:3]
for i in t3:
    print(students[i], "  ", marks[i])
print()

# 5. Display the Lowest 3 students based on marks.
l3 = ar[-1:-4:-1]
for i in l3:
    print(students[i], "  ", marks[i])
```

```python
# Find all students who scored more than 70 marks.
ran = marks > 70
for i in np.where(ran)[0]:
    print(students[i], marks[i])
print()

# Find all students who scored less than 50 marks.
ran1 = marks < 50
for i in np.where(ran1)[0]:
    print(students[i], marks[i])
    print()

# Use np.argwhere() to find the indices of students who scored more than 80 marks. 
index = np.argwhere(marks > 80)
print(index)
print()

# Use np.nonzero() to find the indices of non-zero marks. 
index = np.nonzero(marks)
print(index)
print()

# Use np.count_nonzero() to count the number of students who scored more than 60 marks. 
count = np.count_nonzero(marks > 60)
print("Number of students:", count)
print()

# Sort the marks and use np.searchsorted() to find the position where 70 can be inserted while 
# maintaining sorted order. 
sorted = np.sort(marks)
print(sorted)
print()

# Use np.isin() to find students whose marks are: 
# 45, 67, 95 
required_marks = [45, 67, 95]
result = np.isin(marks, required_marks)
print(result)
print()
```

```python
# A student is considered Pass if marks are greater than or equal to 40. 
# Use np.where() to create the result: 
# Pass 
# Fail 
result = np.where(marks >= 40, "Pass", "Fail")
print(result)
print()
```

```python
# Use Boolean indexing to display students who passed. 
p = marks >= 40
print(students[p])
```

```python
# Display students who scored less than 40.
f = marks < 40

print(students[f])
```

```python
# Use np.select() to assign grades according to the following rules: 
# Marks Grade 
# < 40 Fail 
# 40 – 59 C 
# 60 – 74 B 
# 75 – 89 A 
# 90 – 100 A+ 
# Display: 
# Student Name 
# Marks 
# Grade 
marks = np.array([78, 45, 92, 67, 35, 88, 56, 74, 95, 62])
conditions = [
    marks < 40,
    (marks >= 40) & (marks <= 59),
    (marks >= 60) & (marks <= 74),
    (marks >= 75) & (marks <= 89),
    (marks >= 90) & (marks <= 100)
]

grades = ["Fail", "C", "B", "A", "A+"]
result = np.select(conditions, grades, default="Unknown")
print(result)

for i in range(len(students)):
    print(students[i], marks[i], result[i])
```

```python
# Create another marks array containing some invalid values: 
# 78, 45, 110, 67, -5, 88 
# Marks should always be between 0 and 100. 
# Use np.clip() to convert the values into the valid range. 
# Expected rule: 
# Below 0  → 0 
# 0–100    → unchanged 
# Above 100 → 100 
new = np.array([78, 45, 110, 67, -5, 88])
valid = np.clip(new, 0, 100)
print(valid)
```

```python
# Find students whose marks are: 
# >= 60 AND <= 80 
# Use: 
# np.logical_and() 
result = np.logical_and(marks >= 60, marks <= 80)
print(students[result])
print(marks[result])
```

```python
 # Find students whose marks are: 
# < 40 OR > 90 
# Use: 
# np.logical_or()
result = np.logical_or(marks < 40, marks > 90)
print(students[result])
print(marks[result])
```

```python
# 1. Total marks  
total = np.sum(marks)
print("Total marks:", total)
print()

# 2. Average marks  
average = np.mean(marks)
print("Average marks:", average)
print()

# 3. Median marks 
median = np.median(marks)
print("Median marks:", median)
print()

# 4. Minimum marks  
minimum = np.min(marks)
print("Minimum marks:", minimum)
print()

# 5. Maximum marks 
maximum = np.max(marks)
print("Maximum marks:", maximum)
print()
```

```python
# 6. Standard deviation  
sd = np.std(marks)
print("Standard deviation:", sd)
print()

# 7. Variance  
v = np.var(marks)
print("Variance:", v)

print()

# 8. 25th percentile  
p_25 = np.percentile(marks, 25)
print("25th percentile:", p_25)
print()

# 9. 50th percentile  
p_50 = np.percentile(marks, 50)
print("50th percentile:", p_50)
print()

# 10. 75th percentile  
p_75 = np.percentile(marks, 75)
print("75th percentile:", p_75)
print()
```

```python
# 11. Index of the highest marks  
hindex = np.argmax(marks)
print("Index of highest marks:", hindex)
print()

# 12. Index of the lowest marks  
lindex = np.argmin(marks)
print("Index of lowest marks:", lindex)
print()

# 13. Name of the highest-scoring student  
hstudent = students[hindex]
print("Highest scoring student:", hstudent)
print()

# 14. Name of the lowest-scoring student  
lstudent = students[lindex]
print("Lowest-scoring student:", lstudent)
```

```python
# 8. Find the maximum while ignoring missing values.  
maximum = np.nanmax(marks)
print("Maximum:", maximum)
```

```python
# Create a random number generator using:
rng = np.random.default_rng(42)
```

```python
# Generate 10 random student marks between 1 and 100 using: 
# rng.integers()
random_marks = rng.integers(1, 101, size=10)
print(random_marks)
```

```python
# Generate 5 random floating-point values between 0 and 1 using: 
# rng.random()
random_values = rng.random(5)
print(random_values)
```

```python
# Create the following course array: 
# Python 
# Java 
# Data Science 
# Data Analytics 
# Randomly select 5 courses using: 
courses = np.array(['Python','Java','Data Science','Data Analytics'])
random_courses = rng.choice(courses, size=5)
print(random_courses)
```

```python
# Randomly select 3 students from the student array. 
# The same student should not be selected more than once.
random_students = rng.choice(students, size=3, replace=False)
print(random_students)
```

```python
# Generate 10 random values using a normal distribution with: 
# Mean = 60 
# Standard Deviation = 10
normal_values = rng.normal(60, 10, size=10)
print(normal_values)
```

```python
# Generate 5 random values between: 
# 10 and 20
uniform_values = rng.uniform(10, 20, size=5)
print(uniform_values)
```

```python
# Simulate 10 experiments where: 
# Number of trials = 10 
# Probability of success = 0.5 
binomial_values = rng.binomial(10, 0.5, size=10)
print(binomial_values)
```

```python
# Generate 10 random values using a Poisson distribution where: 
# Lambda = 3
poisson_values = rng.poisson(3, size=10)
print(poisson_values)
```

```python
# Create a copy of the student array and randomly shuffle it using: 
# rng.shuffle() 
# Make sure the original student array remains unchanged.
shuffled_students = students.copy()
rng.shuffle(shuffled_students)
print(students)
print("Shuffled Students:",shuffled_students)
```

```python
# Create a random permutation of the student array using: 
# rng.permutation() 
# Display both: 
# Original Students 
# Permuted Students 
permuted_students = rng.permutation(students)
print("Original Students:",students)
print("Permuted Students:",permuted_students)
```
