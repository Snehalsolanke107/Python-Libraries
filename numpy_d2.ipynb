```python
# math
a1 = np.array([5,6,7,8])
a2 = np.array([1,2,3,4])
np.subtract(a1,a2)
np.divide(a1,a2)
np.add(a1,a2)
np.divide(a1,a2)
np.power(a1,a2)
np.sqrt(a1)
np.square(a1)
np.abs(a1,a2) # convert -values into +-ve
np.sqrt(a2)
np.square(a2)
```

```python
#std
a1 = np.array([5,6,7,8])
np.min(a1)
np.max(a1)
np.median(a1)
np.std(a1)
np.var(a1)
```

```python
#copy & view
a1 = np.array([1, 2, 3, 4, 5])
a2 = a1.copy()
print(a1)
print(a2)
a2[-1] = 77
print(a2)
print(a1)
```

```python
a1 = np.array([1, 2, 3, 4, 5])
a2 = a1.view()
print(a1)
print(a2)
a2[-1] = 77
print(a2)
print(a1)
```

```python
a1 = np.array([1,2,3,4])
a2 = np.array([5,6,7,8])
# hstack, vstack
ah = np.hstack((a1,a2))
print(ah)

av = np.vstack((a1,a2))
print(av)
```
