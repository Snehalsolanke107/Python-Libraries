```python
import pandas as pd
```

```python
import numpy as np
```

```python
df = pd.read_excel("E-Commerce Sales.xlsx")
df
```

```python
# Check shape, info() and describe().  

print(df.shape)
print()
df.info()
print()
print(df.describe())
```

```python
# Display column names. 
print("Columns:", df.columns)
```

```python
# Check unique categories and cities. 
print("Categories:", df["Category"].unique())
print("Cities:", df["City"].unique())
```

```python
# Check the number of unique customers.  
print("Unique customers:", df["Customer_Name"].nunique())
```

```python
# Identify missing values. 
print(df.isnull().sum())
```

```python
# Handle missing Discount and Rating appropriately.
print(df["Discount"].fillna(0))
print()
median = df["Rating"].median()
df["Rating"] = df["Rating"].fillna(median)
print(df["Rating"])
```

```python
print(df[["Discount", "Rating"]].isnull().sum())
```

```python
# Find and remove duplicate records.  
print(df[df.duplicated()])
print(df.duplicated().sum())
print()
df = df.drop_duplicates()
print(df)
```

```python
# Check whether Order_ID contains duplicates. 
print(df[df["Order_ID"].duplicated(keep=False)])
```

```python
# Convert Order_Date into proper datetime format. 
df["Order_Date"] = pd.to_datetime(df["Order_Date"], errors="coerce")
print(df["Order_Date"].dtype)
```

```python
# Gross_Amount = Quantity × Unit_Price 
df["Gross_Amount"] = df["Quantity"] * df["Unit_Price"]
df
```

```python
# Discount_Amount = Gross_Amount × Discount / 100
df["Discount_Amount"] = (df["Gross_Amount"] * df["Discount"] / 100)
df
```

```python
# Net_Sales = Gross_Amount - Discount_Amount 
df["Net_Sales"] = (df["Gross_Amount"] - df["Discount_Amount"])
df
```

```python
# Top 5 orders based on Net_Sales. 
print(df.sort_values("Net_Sales", ascending=False).head(5))
```

```python
# Total sales generated. 
total_sales = df["Net_Sales"].sum()

print("Total Sales Generated: ₹", total_sales)
```

```python
# Average order value. 
print(df["Net_Sales"].mean())
```

```python
# Highest and lowest order value. 
print(df.loc[df["Net_Sales"].idxmax()])
print(df.loc[df["Net_Sales"].idxmin()])
```

```python
# Most frequently purchased product. 
print(df["Product"].value_counts().head(1))
```

```python
# Most common category. 
print(df["Category"].value_counts().head(1))
```

```python
# Customers who have placed orders above ₹20,000. 
print(df.loc[df["Net_Sales"] > 20000,["Customer_Name", "Order_ID", "Net_Sales"]])
```

```python
# Orders where discount is greater than 15%.  
print(df.loc[df["Discount"] > 15,["Order_ID", "Product", "Discount"]])
```

```python
# Orders having rating below 3. 
print(df.loc[df["Rating"] < 3,["Order_ID", "Customer_Name", "Product", "Rating"]])
```

```python
# Convert Net_Sales into a NumPy array and calculate: 
sales = df["Net_Sales"].dropna().to_numpy()
# • Mean sales  
mean_sales = np.mean(sales)
print("Mean Sales:", mean_sales)
# • Median sales  
median_sales = np.median(sales)
print("Median Sales:", median_sales)
# • Maximum sales  
maximum_sales = np.max(sales)
print("Maximum Sales:", maximum_sales)
# • Minimum sales  
minimum_sales = np.min(sales)
print("Minimum Sales:", minimum_sales)
# • Standard deviation 
std_sales = np.std(sales)
print("Standard Deviation:", std_sales)
```

```python
# Also create a new column: 
# Sales_Status 
# using np.where(): 
# Net_Sales >= Average Sales → High Value 
# Otherwise                  → Regular

average_sales = df["Net_Sales"].mean()
df["Sales_Status"] = np.where(df["Net_Sales"] >= average_sales,"High Value","Regular")

print(df[["Order_ID", "Net_Sales", "Sales_Status"]])
```

```python
# Sort the data by Net_Sales descending.  
df = df.sort_values(by="Net_Sales", ascending=False)
print(df)
```

```python
# Keep only the required business columns.  
df = df[["Order_ID","Order_Date","Customer_Name","City","Category","Product","Quantity","Unit_Price","Discount","Rating","Gross_Amount","Discount_Amount","Net_Sales","Sales_Status"]]
df
```

```python
# Save the final cleaned data as:
df.to_excel("E-Commerce_Sales_Analysis.xlsx", index=False)
df
```
