```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
sales = [10000,20000,23000,28000,35000]
year = [2021,2022,2023,2024,2025]
plt.plot(year,sales)
plt.show()
```

```python
students = {
    'Month' : ['jan','feb','march','april'],
    'st_no' : [30,45,20,65]
}
student_df = pd.DataFrame(students)
student_df
```

```python
month = student_df['Month']
count = student_df['st_no']
plt.plot(month,count)
plt.show()
```

```python
plt.title('Students Data')
plt.xlabel('Months wise addmission')
plt.ylabel('Count of student')
plt.plot(month,count)
plt.grid()
plt.show()
```

```python
y = [100,200,300,400]
x = [10,15,20,40]
plt.xticks([10,20,30,40,50,60])
plt.yticks([100,200,300,400,500])
plt.plot(x,y)
```

```python
plt.title('Students Data')
plt.xlabel('Months')
plt.ylabel('Count of student')
plt.plot(month,count)
plt.grid()
plt.show()
```

```python
plt.plot(month,count,color="red",linewidth=3)
plt.title('Students Data')
plt.xlabel('Months')
plt.ylabel('Count of student')
plt.plot(month,count)
plt.grid()
plt.show()
```

```python
plt.plot(month,count,color='red',linewidth=2,linestyle='-.')
plt.title('Students Data')
plt.xlabel('Months')
plt.ylabel('Count of student')
plt.plot(month,count)
plt.grid()
plt.show()
```

```python
plt.plot(month,count,color='red',linewidth=2,linestyle='--',marker = 'o',ms=5,mfc='green',mec='black')
plt.title('Students Data')
plt.xlabel('Months')
plt.ylabel('Count of student')
plt.plot(month,count)
plt.grid()
plt.show()
```

```python
plt.plot(month,count,color='red',linewidth=2,linestyle='-.',marker = 'o',ms=10,mfc='green',mec='black')
plt.title('Students Data',loc='left')
plt.xlabel('Months',loc='right')
plt.ylabel('Count of student',loc='center')
plt.plot(month,count)
plt.grid()
plt.show()
```

```python
plt.plot(month,count,color='red',linewidth=2,linestyle='-.',marker = 'o',ms=10,mfc='green',mec='black')
title_font = {
    'family' : 'serif',
    'color' : 'darkred',
    'weight' : 'bold',
    'size' : 16
}

plt.title('Students Data',loc='left',fontdict=title_font)
plt.xlabel('Months',loc='right',fontdict=title_font)
plt.ylabel('Count of student',fontdict=title_font)
plt.plot(month,count)
plt.grid()
plt.show()
```
