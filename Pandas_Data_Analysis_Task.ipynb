```python
department_summary = df.groupby("Department").agg(
    Employee_Count=("Employee_ID", "count"),
    Average_Salary=("Monthly_Salary", "mean"),
    Average_Performance=("Performance_Score", "mean"),
    Average_Attendance=("Attendance_Percentage", "mean"),
    Total_Tasks_Assigned=("Tasks_Assigned", "sum"),
    Total_Tasks_Completed=("Tasks_Completed", "sum")
)

print(department_summary)
```

```python
highest = department_summary["Average_Performance"].idxmax()

print("Highest Average Performance Department:", highest)
```

```python
def performance_category(score):
    if score >= 85:
        return "Excellent"
    elif score >= 70:
        return "Good"
    elif score >= 50:
        return "Average"
    else:
        return "Needs Improvement"
```

```python
df["Performance_Category"] = df["Performance_Score"].apply(performance_category)
df
```

```python
def employee_status(row):
    if (row["Attendance_Percentage"] >= 90 and
        row["Task_Completion_Percentage"] >= 90 and
        row["Performance_Score"] >= 80):
        
        return "Top Performer"

    elif (row["Attendance_Percentage"] >= 75 and
          row["Task_Completion_Percentage"] >= 70 and
          row["Performance_Score"] >= 60):
        
        return "Satisfactory"

    else:
        return "Needs Attention"
```

```python
df["Employee_Status"] = df.apply(employee_status, axis=1)
df
```

```python
top_performers = (df["Employee_Status"] == "Top Performer").sum()
dept_salary = df.groupby("Department")["Monthly_Salary"].mean()
highest_salary_department = dept_salary.idxmax()
```

```python
dept_attendance = df.groupby("Department")["Attendance_Percentage"].mean()
best_attendance_department = dept_attendance.idxmax()
```

```python
highest_score = df["Performance_Score"].max()
employee = df[df["Performance_Score"] == highest_score]
print(employee[["Employee_ID","Employee_Name","Performance_Score"]])
```

```python
highest_completion = df["Task_Completion_Percentage"].max()
employee = df[df["Task_Completion_Percentage"] == highest_completion]
print(employee[["Employee_ID","Employee_Name","Task_Completion_Percentage"]])
```

```python
needs_attention = (df["Employee_Status"] == "Needs Attention").sum()
percentage = needs_attention / len(df) * 100
print("Needs Attention:", round(percentage, 2), "%")
```
