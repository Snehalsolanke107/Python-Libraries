```python
# Check the dataset for missing/null values.
print(student_df.isnull())
```

```python
# Find the total number of null values present in each column.
print(student_df.isnull().sum())
```

```python
# Replace missing values:  
# • Missing City → "Unknown" 
student_df["City"] = student_df["City"].fillna("Unknown")
student_df
```

```python
# Missing Course → "Not Assigned"  
student_df["Course"] = student_df["Course"].fillna("Not Assigned")
student_df
```

```python
# Missing Email → "Not Available"
student_df["Email"] = student_df["Email"].fillna("Not Available")
student_df
```

```python
# Missing Placement_Status → "Not Updated" 
student_df["Placement_Status"] = student_df["Placement_Status"].fillna("Not Updated")
student_df
```

```python
# Replace missing values in:  
# • Marks using the average marks  
average_marks = student_df["Marks"].mean()
student_df["Marks"] = student_df["Marks"].fillna(average_marks)
print(student_df["Marks"])
```

```python
# • Fees using the median fees  
median_fees = student_df["Fees"].median()
student_df["Fees"] = student_df["Fees"].fillna(median_fees)
print(student_df["Fees"])
```

```python
# • Attendance using the average attendance  
average_attendance = student_df["Attendance"].mean()
student_df["Attendance"] = student_df["Attendance"].fillna(average_attendance)
print(student_df["Attendance"])
```

```python
# Create a separate copy of the original DataFrame and remove 
# rows containing null values. 
student_copy = student_df.copy()
student_copy = student_copy.dropna()
print(student_copy)
print("Shape after removing null values:", student_copy.shape)
```

```python
# . Check whether duplicate records are present in the dataset.
duplicates = student_df.duplicated()
print(duplicates)
print("Total duplicate records:", duplicates.sum())
```

```python
# Identify and display the duplicate records. 
duplicate = student_df[student_df.duplicated()]
duplicate
```

```python
# Remove the duplicate records from the dataset. 
student_df = student_df.drop_duplicates()
student_df
```

```python
# Delete the Email column from the DataFrame. 
student_df = student_df.drop("Email", axis=1)
student_df
```

```python
# Delete the Attendance column from the DataFrame. 
student_df = student_df.drop("Attendance", axis=1)
student_df
```

```python
# Delete:  
# • First row  
df1 = student_df.drop(index=0)
print(df1)
# • Last row  
df2 = student_df.drop(index=student_df.index[-1])
print(df2)
# • Row at index 5  
df3 = student_df.drop(index=5)
print(df3)
# • Rows at indexes 2, 4, and 7
df4 = student_df.drop(index=[2, 4, 7])
print(df4)
```

```python
# Create a final cleaned dataset by:  
# • Removing duplicate records  
student_df = student_df.drop_duplicates()
student_df
# • Handling all missing values  
student_df["City"] = student_df["City"].fillna("Unknown")
student_df["Course"] = student_df["Course"].fillna("Not Assigned")
student_df["Email"] = student_df["Email"].fillna("Not Available")
student_df["Placement_Status"] = student_df["Placement_Status"].fillna("Not Updated")
student_df
```

```python
# • Removing unnecessary columns 
student_df = student_df.drop(
    columns=["Email", "Attendance"])
student_df
```

```python
# After cleaning, verify that:  
# • No unwanted null values remain  
print(student_df.isnull().sum())
```
