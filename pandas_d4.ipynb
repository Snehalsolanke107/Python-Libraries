```python
df = pd.read_excel("Pandas_Merge_Practice.xlsx",sheet_name=None)
df.keys()
```

```python
Customers_df = df['Customers']
Customers_df
```

```python
Orders_df = df['Orders']
Orders_df
```

```python
inner_df = pd.merge(Customers_df,Orders_df,how='inner',on='Customer_ID')
inner_df
```

```python
Left_df = pd.merge(Customers_df,Orders_df,how='left',on='Customer_ID')
Left_df
```

```python
Right_df = pd.merge(Customers_df,Orders_df,how='right',on='Customer_ID')
Right_df
```

```python
Outer_df = pd.merge(Customers_df,Orders_df,how='outer',on='Customer_ID')
Outer_df
```
