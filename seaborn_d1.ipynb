```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
sns.get_dataset_names()
```

```python
tips_df = sns.load_dataset('tips')
tips_df
```

```python
sns.barplot(data = tips_df,x='sex',y='total_bill',estimator='sum')
```

```python
sns.barplot(data = tips_df,x='sex',y='total_bill',estimator='sum',hue='time')
plt.title('Total bill by gender')
plt.xlabel('Gender')
plt.ylabel('Bill amount')
plt.show()
```

```python
sns.barplot(data = tips_df,x='sex',y='total_bill',estimator='sum',hue='time',gap=0.1,color='red')
plt.title('Total bill by gender')
plt.xlabel('Gender')
plt.ylabel('Bill amount')
plt.show()
```

```python
sns.scatterplot(data = tips_df,x='total_bill',y='tip',hue='sex',style='time',size='tip')
```

```python
sns.countplot(data=tips_df,x='sex',hue='time',color='green')
```
